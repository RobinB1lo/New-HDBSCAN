#### A few notes:
- Our monte carlo will always be 1000 sims
- Our noise is added random gaussian noise to the dataset
- We are using a probabaility rank matrix where each entry (i, j) is "the probability that there are at least k closer points to i than j is", because our second matrix with the entries "probabaility that j is in the k nearest neighbours of i" only works for datasets with very little noise
- We will be doing 3 trials for our probability rank matrix simulation and vanilla HDBSCAN, because of the randomness of our noise (we want to make sure we didn't just get lucky) (# of trials might change in the future)
- The standard deviation of our dataset will stay constant at 0.5

```python
def avg(*args: float) -> float:
    sum = 0
    for num in args:
        sum += num
    return sum/len(args)

print(avg(0.027200123639856785, 0.001500845841593072, 0.006105120246615588))
```

```python
import matplotlib.pyplot as plt

def plot(vanilla: list[float], prob: list[float], num_blobs: int) -> None:
    noise_levels = [1.0, 2.0, 3.0]
    plt.figure(figsize=(6,4))
    plt.plot(noise_levels, vanilla, marker='o', linestyle='-', label='Vanilla HDBSCAN')
    plt.plot(noise_levels, prob, marker='s', linestyle='--', label='Uncertainty aware')
    plt.xlabel('Observation noise')
    plt.ylabel('Adjusted Rand Index (ARI)')
    plt.title(f'ARI vs. Noise Level ({num_blobs} Gaussian Blobs)')
    plt.xticks(noise_levels)  
    plt.ylim(-0.5,1.5)              
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

#Plotting vanilla hdbscan against our hdbscan with a probability rank matrix with increasing amount of noise and blobs

# 4 blobs
vanilla_4 = [0.9775864561092047, 0.12938813382299588, 0.0013079208276172182]
prob_4 = [0.9701817503576393, 0.28882546906636697, 0.024130497671845812]
plot(vanilla_4, prob_4, 4)

# 5 blobs
vanilla_5 = [0.6543201982212593, 0.05162258230756939, 0.02281883659864585]
prob_5 = [0.7045087188564795, 0.06506755597453366, 0.00985038469319393]
plot(vanilla_5, prob_5, 5)

# 6 blobs
vanilla_6 = [0.8390292517453205, 0.01605562718270633, 0.003725628209360244]
prob_6 = [0.8089681832992165, 0.10637255530557166, 0.01144266061781792]
plot(vanilla_6, prob_6, 6)
```
