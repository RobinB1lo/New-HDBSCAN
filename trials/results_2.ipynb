### Will use to keep track of tests of our second implementation of the probabaility rank matrix which sets all values under a certain benchmark to 0
- Our monte carlo will always be 1000 sims
- Our noise is added random gaussian noise to the dataset
- We are using a probabaility rank matrix where each entry (i, j) is "the probability that there are at least k closer points to i than j is" with our certainty threshold, because our second matrix with the entries "probabaility that j is in the k nearest neighbours of i" only works for datasets with very little noise
- We will be doing 3 trials for our probability rank matrix simulation and vanilla HDBSCAN, because of the randomness of our noise (we want to make sure we didn't just get lucky) (# of trials might change in the future)
- The standard deviation of our dataset will stay constant at 0.5

```python
def avg(*args: float) -> float:
    sum = 0
    for num in args:
        sum += num
    return sum/len(args)
```

```python
import matplotlib.pyplot as plt

def plot(vanilla: list[float], prob: list[float], num_blobs: int) -> None:
    noise_levels = [1.0, 2.0, 3.0]
    plt.figure(figsize=(6,4))
    plt.plot(noise_levels, vanilla, marker='o', linestyle='-', label='Vanilla HDBSCAN')
    plt.plot(noise_levels, prob, marker='s', linestyle='--', label='Uncertainty aware')
    plt.xlabel('Observation noise')
    plt.ylabel('Adjusted Rand Index (ARI)')
    plt.title(f'ARI vs. Noise Level ({num_blobs} Gaussian Blobs)')
    plt.xticks(noise_levels)  
    plt.ylim(-0.5,1.5)              
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```
